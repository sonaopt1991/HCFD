Chi-Square Test for Independence For feature importance

```python

from scipy.stats import chi2_contingency
columns_to_test = ['BeneID', 'ClaimID', 'ClaimStartDt', 'ClaimEndDt','Provider']

for col in columns_to_test:

    crosstab = pd.crosstab(data[col], data['PotentialFraud'])

    # Perform Chi-Square test
    chi2, p, dof, expected = chi2_contingency(crosstab)

    print(f"Chi-square test for {col}:")
    print(f"Chi2: {chi2}, p-value: {p}")

    #  significance level (alpha)
    alpha = 0.05
    if p < alpha:
        print(f"There is a statistically significant association between {col} and PotentialFraud.")
    else:
        print(f"There is no statistically significant association between {col} and PotentialFraud.")
    print("-" * 20)
```

```python
def calculate_overlap(row):
    # Create a set of non-null physician IDs
    physicians = {row['AttendingPhysician'], row['OperatingPhysician'], row['OtherPhysician']}
    physicians.discard(None)  # Remove None or NaN values from the set

    # Count non-null roles
    non_null_count = sum(pd.notnull([row['AttendingPhysician'], row['OperatingPhysician'], row['OtherPhysician']]))

    # If multiple roles are populated but the same physician is listed, flag as overlap
    if non_null_count > 1 and len(physicians) < non_null_count:
        return 1
    else:
        return 0

# Apply the function to create the PhysicianOverlap feature
data['PhysicianOverlap'] = data.apply(calculate_overlap, axis=1)

```

ANOVA TEST to check the significance of numerical columns with the target column

```python
from scipy.stats import f_oneway
```

```python
# Perform ANOVA test for each numerical column
numerical_columns = data.select_dtypes(include=np.number).columns
target_column = 'PotentialFraud'

anova_results = {}

for col in numerical_columns:
    # Group the numerical column by the target column
    grouped_data = [group[col].values for name, group in data.groupby('PotentialFraud')]

    # Perform one-way ANOVA
    f_stat, p_value = f_oneway(*grouped_data)
    anova_results[col] = {"F-Statistic": f_stat, "p-Value": p_value}

# Display results
for col, results in anova_results.items():
    print(f"ANOVA results for {col}:")
    print(f"  F-Statistic: {results['F-Statistic']:.4f}")
    print(f"  p-Value: {results['p-Value']:.4f}")
    print("  Significant?" + (" Yes" if results['p-Value'] < 0.05 else " No"))
    print()
```

```python
anova_df = pd.DataFrame.from_dict(anova_results, orient='index')
anova_df = anova_df.sort_values(by='p-Value')
anova_df
```

```python
columns_to_remove = [col for col in numerical_columns if anova_results[col]['p-Value'] > 0.05]

# Drop the identified columns from the dataset
data = data.drop(columns=columns_to_remove, errors='ignore')
```

```python
# Calculate global mean
global_mean = data['PotentialFraud'].mean()

# Function for applying smoothing
def smooth_encoding(group, global_mean, smoothing_factor=10):
    mean_target = group['PotentialFraud'].mean()
    count = group['PotentialFraud'].count()
    return (mean_target * count + global_mean * smoothing_factor) / (count + smoothing_factor)

# Step 1: Group by 'BeneID' and apply smoothing
target_encoding = data.groupby('BeneID').apply(lambda x: smooth_encoding(x, global_mean)).reset_index(name='TargetEncoded')

# Step 2: Merge the target encoding back into the original DataFrame
data= data.merge(target_encoding, on='BeneID', how='left')


```
